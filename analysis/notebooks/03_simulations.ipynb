## Load models and media

```python
import os
from cobra.io import read_sbml_model
import pandas as pd


def set_culture_media(model, media):
    """Apply the media bounds, closing any `_reverse` import partner.

    GECKO models bypass the medium cap entirely if those partners are left open: ecBSU1 was
    taking up 9.3 mmol/gDW/h of ammonia against an M9 cap of 5.
    """
    for _, row in media.iterrows():
        reaction_id = row['ID']
        if reaction_id not in model.reactions:
            continue
        model.reactions.get_by_id(reaction_id).bounds = (row['LOWER_BOUND'], row['UPPER_BOUND'])

        reverse_id = f"{reaction_id}_reverse"
        if reverse_id in model.reactions:
            model.reactions.get_by_id(reverse_id).bounds = (0.0, 0.0)
    return model


DATASETS_DIR = os.path.join("..", "data", "datasets")
TABLES_DIR = os.path.join("..", "results", "tables")
os.makedirs(TABLES_DIR, exist_ok=True)

medium = pd.read_csv(os.path.join(DATASETS_DIR, "M9_media.csv"))
final_dir = os.path.join("..", "data", "models", "02_final_annotated")
model_names = ["iYO844", "iBsu1103", "iBsu1103v2", "iBsu1147", "eciYO844", "iBB1018", "iBsu1147R", "ecBSU1", "iBsu1209", "Submodel"]
modelos = {name: read_sbml_model(os.path.join(final_dir, f"{name}.xml")) for name in model_names}
```

## Run experiments 1 & 2

Per condition, closes default glucose uptake and applies each row's `reactions`/`import fluxes`
as lower bounds, then optimizes. Writes `Chubukov_2013_results`/`Dauner_2001_results` sheets.

```python
def run_experiment(model, experiment_data):

    simulated_growth = []

    for _, row in experiment_data.iterrows():    

        reactions = [r.strip() for r in str(row["reactions"]).split(";")]
        fluxes = [float(f.strip()) for f in str(row["import fluxes"]).split(";")]

        with model:
            model.reactions.get_by_id("EX_glc__D_e").lower_bound = 0.0
            for rxn_id, flux in zip(reactions, fluxes):
                try:
                    rxn = model.reactions.get_by_id(rxn_id)
                    rxn.lower_bound = flux
                except KeyError:
                    print(f"{model.id}:could not find reaction {rxn_id}")

            solution = model.optimize()

            simulated_growth.append(solution.objective_value)

    return simulated_growth 


# Experiment condition tables live in data/datasets/; the simulated results they produce are
# pipeline output and go to results/tables/, one CSV per experiment.
EXPERIMENTS = {"Chubukov_2013": "Chubukov_2013.csv", "Dauner_2001": "Dauner_2001.csv"}

for experiment_name, experiment_file in EXPERIMENTS.items():

    experiment = pd.read_csv(os.path.join(DATASETS_DIR, experiment_file))
    results = experiment.copy()

    for model_name, model in modelos.items():
        model_m9 = set_culture_media(model, medium)
        results[model_name] = run_experiment(model_m9, experiment)

    out_path = os.path.join(TABLES_DIR, f"{experiment_name}_results.csv")
    results.to_csv(out_path, index=False)
    print(f"[INFO] {len(results)} rows written to {out_path}")
```

## Dauner, 2001 with caa3 oxidase knocked out

Re-runs experiment 2 with `CYOO3` (cytochrome caa3 oxidase) knocked out where present - the
literature suggests this better represents glucose-fed conditions. Models without a `CYOO3`
reaction (`iBsu1103`, `iBsu1103v2`, `eciYO844`, `iBsu1209`) run unmodified.

```python
experiment = pd.read_csv(os.path.join(DATASETS_DIR, "Dauner_2001.csv"))
results = experiment.copy()

for model_name, model in modelos.items():

    model_m9 = set_culture_media(model, medium)

    if "CYOO3" in model_m9.reactions:
        print(f"{model_name}: CYOO3 found, knocking out")
        model_m9.reactions.get_by_id("CYOO3").bounds = (0, 0)
    else:
        print(f"{model_name}: no CYOO3")

    results[model_name] = run_experiment(model_m9, experiment)

out_path = os.path.join(TABLES_DIR, "Dauner_2001caa3KO_results.csv")
results.to_csv(out_path, index=False)
print(f"\n[INFO] {len(results)} rows written to {out_path}")
```

## Terminal oxidase usage, before and after the knockout

Which terminal oxidase each model uses to reduce oxygen, under M9 with glucose, before and after
`CYOO3` is knocked out. Writes `../results/tables/terminal_oxidase_flux.csv` (supplementary
Table S3).

Fluxes are summed over every reaction whose id starts with the given prefix, so the GECKO isozyme
copies in `ecBSU1` are aggregated. Any remaining oxygen-consuming reaction outside the three named
oxidases is collected in `other`, which is what the Submodel falls back on.

```python
OXIDASES = {"caa3": "CYOO3", "aa3_600": "CYTB_B3", "bd": "CYTB_B2"}
FUNCTIONAL = ["iYO844", "iBsu1147", "eciYO844", "iBB1018", "iBsu1147R", "ecBSU1", "Submodel"]


def oxidase_fluxes(model, solution):
    """Flux per terminal oxidase, plus any other oxygen-consuming reaction."""
    named = {label: sum(solution.fluxes[r.id] for r in model.reactions
                        if r.id.startswith(prefix))
             for label, prefix in OXIDASES.items()}
    o2 = [m for m in model.metabolites if m.id.startswith("o2_")]
    consumers = {r.id for m in o2 for r in m.reactions
                 if any(x.id.startswith("o2_") and c < 0 for x, c in r.metabolites.items())}
    accounted = {r.id for r in model.reactions
                 if any(r.id.startswith(p) for p in OXIDASES.values())}
    named["other"] = sum(solution.fluxes[rid] for rid in consumers - accounted
                         if abs(solution.fluxes[rid]) > 1e-9)
    return named


records = []
for model_name in FUNCTIONAL:
    # Reloaded rather than taken from `modelos`: the knockout cell above sets CYOO3 bounds on the
    # shared objects outside a `with` block, so those models stay knocked out afterwards.
    model = set_culture_media(
        read_sbml_model(os.path.join(final_dir, f"{model_name}.xml")), medium)

    solution = model.optimize()
    row = {"Model": model_name, "condition": "wild type", "mu": solution.objective_value}
    row.update(oxidase_fluxes(model, solution))
    records.append(row)

    with model:
        for reaction in [r for r in model.reactions if r.id.startswith("CYOO3")]:
            reaction.knock_out()
        solution = model.optimize()
        row = {"Model": model_name, "condition": "caa3 knockout",
               "mu": solution.objective_value}
        row.update(oxidase_fluxes(model, solution))
        records.append(row)

oxidase_table = pd.DataFrame(records).set_index(["Model", "condition"])
out_path = os.path.join(TABLES_DIR, "terminal_oxidase_flux.csv")
oxidase_table.to_csv(out_path)
print(f"[INFO] written to {out_path}\n")
print(oxidase_table.round(2).to_string())
```
